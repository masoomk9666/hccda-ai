```python
def deposit(balance,amount):
    balance =balance+amount
    return balance

def withdraw (balance,amount):
    if amount < balance:
        print("Your balance is Insufficient")
    else:
        balance=balance-amount
    return 

def check (balance):
    print("your balance is:", balance)
balance=10000
while True:
    print('\n----ATM Menu----')
    print("1.Deposite")
    print("2.Withdraw")
    print("3.Check Balance")
    print("4.Exit")
    choice = input("enter your choice:")
    if choice == "1":
        amount=int(input("enter deposit amount: "))
        balance= deposit(balance,amount)
        print(f"amount deposit succesfully and Updated balance is {balance}")
    elif choice=="2":
        amount=int(input("enter withdrawl amount"))
        balance=withdraw(balance,amount)
        print(f"amount Withdrawl succesfully and Updated balance is {balance}")
    elif choice=="3":
        check(balance)
    elif choice=="4":
        print("Thank you for using ATM: ")
        break
    else:
        print("invalid choice")
```

```python
#Lambda Fucntions

result = lambda marks: "pass" if marks >= 50 else "fail"
print(result(int(input("Enter your Marks"))))
```

```python
discount = lambda price : price + (price * 0.1)
print(discount(int(input("Enter Item original Price"))))
```

```python
c_f = lambda c : (c * 9/5) + 32
print(c_f(int(input("Enter Temprature in Calsius to convert into Fahrenheit"))))
```

```python
name_to_uppercase = lambda name : name.upper()
print(name_to_uppercase(input("Enter Temprature in Calsius to convert into Fahrenheit")))
```

```python
total_price = lambda price, quantity : price * quantity
price = int(input("Enter price of an Item"))
quantity = int(input("Enter Quantity of an Item"))
print(total_price(price, quantity))
```

```python
pos_neg = lambda number: "Positive" if number >= 0 else "Negative"
print(pos_neg(int(input("Enter a Number"))))
```

```python
review_list = ["This IS AWEsome", "Great Products     ", "         Beautiful Product", "This is awesome main"]
lowercase_list = list(map(lambda n : n.lower(), review_list))
left_spaces_clean_list = list(map(lambda n : n.lstrip(), lowercase_list))
right_spaces_clean_list = list(map(lambda n : n.rstrip(), left_spaces_clean_list))

print(right_spaces_clean_list)
```
