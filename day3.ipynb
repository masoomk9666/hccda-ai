```python
print('''
---------Menu--------
|Item   | Price |
|Burger | 500 |
|Frise | 600 |
|Shawarma | 300 |
|Role     | 520 |
|Pizza    | 1500 |
''')

items = {"burger": 500,
         "shawarma" : 300,
        "frise" : 600,
        "role" : 520,
        "pizza" : 1500
           }
selected_items = []
prices = []
while True:
   user_item = input("Enter item to add to your Cart")
   if user_item == "finish":
      print("Done")
      print("You selected", selected_items)
      subtotal = sum(prices)
      gst =  subtotal * 0.05
      total = subtotal + gst
      print('---------- Recipt ------------')
      i=0
      while i < len(selected_items):
           print(f"{selected_items[i]}  {prices[i]}")
           i+=1
      print("-------------")
      print(f"Subtotal : RS.{subtotal}")
      print(f"Tax% : RS.{gst}")
      print(f"Total : RS.{total}")
      break
   elif user_item in items:
      selected_items.append(user_item)
      prices.append(items[user_item])
      print(f"{user_item} is Selected")
   else:
      print(f"{user_item} item not found")
    
```
