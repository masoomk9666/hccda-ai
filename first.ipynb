```python
def load_data():
    try:
        with open("emails.txt","r") as file:
            emails =file.readlines()
            emails =list(map(lambda x:x.strip(),emails))
            return emails
    except:
        print("file not found")
        return[]
def detect_spam(email):
    spam_words =["win","free","100%","meeting"]
    def check(e):
        e= e.lower()
        for word in spam_words:
            if word in e:
                return "spam"
        return "not spam"
    result = list(map(check,email))
    return result
def filter_email(email,label):
    labels = list(filter(lambda x:x[1] == "spam",zip(email,label)))
    return labels
emails =load_data()
print(emails)
label = detect_spam(emails)
print(label)
filter_email(emails,label)
```

```python
# Library Managment System

books = [
    {"Num": 1, "Name": "Namal", "Author": "Nimrah Ahmed", "Eddition": "1st & 2nd"},
    {"Num": 2, "Name": "The Alchemist", "Author": "Paulo Coelho", "Eddition": "25th"},
    {"Num": 3, "Name": "Superintelligence", "Author": "Nick Bostrom", "Eddition": "1st"},
    {"Num": 4, "Name": "The Heart Of Darkness", "Author": "Joseph Conrad", "Eddition": "1st"},
    {"Num": 5, "Name": "Euclid's Elements", "Author": "Euclid", "Eddition": "1st"}
]
#Add Book

def add_books():
    print("\n------ Add Books --------")

    book_num = len(books) + 1
    name = input("Enter a book name: ")
    author = input("Enter an author: ")
    eddition = input("Enter an edition: ")

    book = {
        "Num": book_num,
        "Name": name,
        "Author": author,
        "Eddition": eddition
    }

    books.append(book)

    print("\nBook added successfully!")
    print("Book Number:", book_num)

#View Book

def view_books():
    print("\n------ View Books -------")

    if len(books) == 0:
        print("Book not found.")
        return

    for book in books:
        print("\nBook Number:", book["Num"])
        print("Name:", book["Name"])
        print("Author:", book["Author"])
        print("Edition:", book["Eddition"])

#Search Book

def search_book():
    print("\n---- Search Book -----")

    book_num = int(input("Enter a book number: "))

    found = False

    for book in books:
        if book["Num"] == book_num:
            print("\nBook Found!")
            print("Name:", book["Name"])
            print("Author:", book["Author"])
            print("Edition:", book["Eddition"])

            found = True
            break

    if not found:
        print("Book not found.")

#Delete Book

def delete_book():
    print("\n---- Delete Book ----")

    book_num = int(input("Enter a book number: "))

    for book in books:
        if book["Num"] == book_num:
            books.remove(book)
            print("Book deleted successfully!")
            return

    print("Book not found.")

#Main Menu

while True:

    print("\n==========================")
    print("   Library Management System")
    print("==========================")

    print("1. Add Book")
    print("2. View Books")
    print("3. Search Book")
    print("4. Delete Book")
    print("5. Exit")

    choice = input("\nEnter your choice: ")

    if choice == "1":
        add_books()

    elif choice == "2":
        view_books()

    elif choice == "3":
        search_book()

    elif choice == "4":
        delete_book()

    elif choice == "5":
        print("\nThank You for using Library Management System!")
        break

    else:
        print("\nInvalid Choice. Please try again.")
```
