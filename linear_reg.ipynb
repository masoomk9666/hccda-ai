```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from mpl_toolkits.mplot3d import Axes3D

# Features and target
X = df[['study_hours_per_day', 'attendance_percentage']]
y = df['exam_score']

# Train model
model = LinearRegression()
model.fit(X, y)

# Create mesh grid
x1 = np.linspace(df['study_hours_per_day'].min(),
                 df['study_hours_per_day'].max(), 30)

x2 = np.linspace(df['attendance_percentage'].min(),
                 df['attendance_percentage'].max(), 30)

X1, X2 = np.meshgrid(x1, x2)

# Predict plane values
Z = model.predict(
    np.c_[X1.ravel(), X2.ravel()]
).reshape(X1.shape)

# Plot
fig = plt.figure(figsize=(10,7))
ax = fig.add_subplot(111, projection='3d')

# Scatter points
ax.scatter(df['study_hours_per_day'],
           df['attendance_percentage'],
           df['exam_score'],
           color='blue',
           s=40,
           label='Actual Data')

# Regression Plane
ax.plot_surface(X1, X2, Z,
                alpha=0.5,
                color='red')

# Labels
ax.set_xlabel("Study Hours")
ax.set_ylabel("Attendance Percentage")
ax.set_zlabel("Exam Score")

ax.set_title("Multiple Linear Regression Plane")

plt.show()
```
