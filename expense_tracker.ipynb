```python

FILE_NAME = "expenses.txt"

# Create file if it does not exist
file = open(FILE_NAME, "a")
file.close()


# ==========================================
# ADD EXPENSE
# ==========================================

def add_expense():

    date = input("Enter Date (DD-MM-YYYY): ")
    category = input("Enter Category (Food/Transport/Bills/etc): ")
    amount = input("Enter Amount: ")
    description = input("Enter Description: ")

    file = open(FILE_NAME, "a")

    file.write(date + "," + category + "," + amount + "," + description + "\n")

    file.close()

    print("Expense added successfully!")


# ==========================================
# VIEW ALL EXPENSES
# ==========================================

def view_expenses():

    file = open(FILE_NAME, "r")

    lines = file.readlines()

    file.close()

    if len(lines) == 0:

        print("No expenses found.")

    else:

        print("\n========== ALL EXPENSES ==========")

        for line in lines:

            expense = line.strip().split(",")

            print("Date:", expense[0])
            print("Category:", expense[1])
            print("Amount:", expense[2])
            print("Description:", expense[3])
            print("---------------------------------")


# ==========================================
# SEARCH EXPENSE BY DATE
# ==========================================

def search_expense():

    search_date = input("Enter Date to search (DD-MM-YYYY): ")

    file = open(FILE_NAME, "r")

    found = False

    for line in file:

        expense = line.strip().split(",")

        if expense[0] == search_date:

            print("\nExpense Found!")
            print("Date:", expense[0])
            print("Category:", expense[1])
            print("Amount:", expense[2])
            print("Description:", expense[3])
            print("---------------------------------")

            found = True

    file.close()

    if found == False:

        print("No expenses found for this date.")


# ==========================================
# CALCULATE TOTAL EXPENSE
# ==========================================

def total_expense():

    file = open(FILE_NAME, "r")

    total = 0

    for line in file:

        expense = line.strip().split(",")

        amount = float(expense[2])

        total = total + amount

    file.close()

    print("\nTotal Expense:", total)


# ==========================================
# DAILY TOTAL
# ==========================================

def daily_total():

    search_date = input("Enter Date (DD-MM-YYYY): ")

    file = open(FILE_NAME, "r")

    total = 0
    found = False

    for line in file:

        expense = line.strip().split(",")

        if expense[0] == search_date:

            amount = float(expense[2])

            total = total + amount

            found = True

    file.close()

    if found:

        print("Total Expense on", search_date, "=", total)

    else:

        print("No expenses found for this date.")


# ==========================================
# DELETE EXPENSE
# ==========================================

def delete_expense():

    delete_date = input("Enter Date of expense to delete: ")
    delete_description = input("Enter Description of expense to delete: ")

    file = open(FILE_NAME, "r")

    lines = file.readlines()

    file.close()

    found = False

    file = open(FILE_NAME, "w")

    for line in lines:

        expense = line.strip().split(",")

        if expense[0] == delete_date and expense[3] == delete_description:

            found = True

        else:

            file.write(line)

    file.close()

    if found:

        print("Expense deleted successfully!")

    else:

        print("Expense not found.")


# ==========================================
# MAIN MENU
# ==========================================

while True:
    print("\n====================================")
    print("       DAILY EXPENSE TRACKER")
    print("====================================")
    print("1. Add Expense")
    print("2. View All Expenses")
    print("3. Search Expense by Date")
    print("4. Calculate Total Expense")
    print("5. Calculate Daily Total")
    print("6. Delete Expense")
    print("7. Exit")
    print("====================================")

    choice = input("Enter your choice: ")

    if choice == "1":

        add_expense()

    elif choice == "2":

        view_expenses()

    elif choice == "3":

        search_expense()

    elif choice == "4":

        total_expense()

    elif choice == "5":

        daily_total()

    elif choice == "6":

        delete_expense()

    elif choice == "7":

        print("Thank you for using Daily Expense Tracker!")
        break

    else:

        print("Invalid choice! Please enter 1 to 7.")
```
