## Exercise 5: Calculate Statistics 

### Problem:
Given the array of daily temperatures, calculate:
1. The **mean** (average) temperature
2. The **maximum** temperature
3. The **minimum** temperature
4. The **standard deviation**

```python
temperatures = np.array([72, 75, 78, 82, 79, 73, 76, 80, 77, 74])
```

**Expected Output (approximately):**
```
Mean: 76.6°F
Max: 82°F
Min: 72°F
Std Dev: 3.07°F
```


````python
# Exercise 5: Your Code Here
temperatures = np.array([72, 75, 78, 82, 79, 73, 76, 80, 77, 74])

mean_temp = np.mean(temperatures)
max_temp = np.max(temperatures)
min_temp = np.min(temperatures)
std_dev = np.std(temperatures)

print(f"Mean: {mean_temp:.1f}°F")
print(f"Max: {max_temp}°F")
print(f"Min: {min_temp}°F")
print(f"Std Dev: {std_dev:.2f}°F")
````

## Exercise 12: Normalize Data 

### Problem:
**Normalize** the array below:
- Formula: `(x - mean) / std_deviation`

```python

```

**Expected Output (approximately):**
```
[-1.41 -0.71  0.    0.71  1.41]
```



````python
# Exercise 13: Your Code Here
data = np.array([10, 20, 30, 40, 50])

# Exercise 12: Your Code Here
normalized_data = (data - np.mean(data)) / np.std(data)
print(normalized_data)
````

## Exercise 13: Final Challenge - Student Grade Analysis 

### Problem:
Analyze student performance across 3 subjects:

```python
# Rows = Students (5 students)
# Columns = Subjects (Math, English, Science)
grades = np.array([[85, 90, 78],
                   [92, 88, 95],
                   [78, 85, 80],
                   [90, 92, 88],
                   [88, 86, 91]])
```

### Tasks:
1. Calculate the **average grade per student** (across all subjects)
2. Calculate the **average grade per subject** (across all students)
3. Find the **highest grade** overall
4. Find which **student** has the highest average
5. Count how many grades are **above 85**


````python
# Exercise 15: Your Code Here
grades = np.array([[85, 90, 78],
                   [92, 88, 95],
                   [78, 85, 80],
                   [90, 92, 88],
                   [88, 86, 91]])

# 1. Calculate the average grade per student (across all subjects)
student_avg = np.mean(grades, axis=1)

# 2. Calculate the average grade per subject (across all students)
subject_avg = np.mean(grades, axis=0)

# 3. Find the highest grade overall
highest_grade = np.max(grades)

# 4. Find which student has the highest average
best_student = np.argmax(student_avg)  # returns the 0-based index of the student

# 5. Count how many grades are above 85
grades_above_85 = np.sum(grades > 85)

print("Average per student:", student_avg)
print("Average per subject:", subject_avg)
print("Highest grade overall:", highest_grade)
print("Student with highest average (index):", best_student)
print("Number of grades > 85:", grades_above_85)




````
