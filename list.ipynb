```python
shop=["milk","bread","eggs"]
cart=[]
while True:
    item=input("enter your item")
    if item=="done":
        break
    elif item in shop:
        cart.append(item)
        print("itam added in the cart")
    else:
        print("item not available")

print("Total item in cart:", len(cart))
print("cart:", cart)
```
