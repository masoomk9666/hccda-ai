```python

import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import root_mean_squared_error, r2_score

# 1. Dataset load karein
sk_df = pd.read_csv("./data/data.csv")

# 2. Required columns select karein aur missing values remove karein
df_clean = sk_df[
    ["Departure Delay in Minutes", "Arrival Delay in Minutes"]
].dropna()

# 3. Input (X) aur Output (y)
X = df_clean[["Departure Delay in Minutes"]]
y = df_clean["Arrival Delay in Minutes"]

# 4. Data ko training aur testing mein divide karein
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=2
)

# 5. Model banayein aur sirf training data par train karein
sk_model = LinearRegression()
sk_model.fit(X_train, y_train)

print("Model trained successfully!")
print("Slope:", sk_model.coef_[0])
print("Intercept:", sk_model.intercept_)

# 6. Test data par prediction karein
y_pred = sk_model.predict(X_test)

# 7. Model evaluation
rmse = root_mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

print(f"RMSE: {rmse:.2f}")
print(f"R2 Score: {r2:.4f}")

# 8. Graph ke liye predictions
# X_train ko sort karein taake line sahi draw ho
train_pred = sk_model.predict(X_train)
plot_df = pd.DataFrame({
    "Departure": X_train["Departure Delay in Minutes"],
    "Arrival": train_pred
}).sort_values("Departure")

plt.figure(figsize=(10, 6))

plt.scatter(
    X_test["Departure Delay in Minutes"],
    y_test,
    color="blue",
    alpha=0.4,
    label="Actual test data"
)

plt.plot(
    plot_df["Departure"],
    plot_df["Arrival"],
    color="red",
    linewidth=2,
    label="Regression line"
)

plt.xlabel("Departure Delay in Minutes")
plt.ylabel("Arrival Delay in Minutes")
plt.title("Departure Delay vs Arrival Delay")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()

# 9. Naye data par prediction
new_data = pd.DataFrame({
    "Departure Delay in Minutes": [30]
})

predicted_arrival = sk_model.predict(new_data)

print(
    "Predicted arrival delay for 30 minutes departure delay:",
    round(predicted_arrival[0], 2),
    "minutes"
)
```

```python
sk_model.fit(X_train_scaled, y_train)
```

```python
sk_model.predict(X_test_scaled)
```
