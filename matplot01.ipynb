```python
batsman = pd.read_csv(r'C:\Users\CorvitsystemISB\Desktop\Saad Ai\data\pakistani_cricketers_scores.csv')
batsman.head()
bat = batsman.copy()
bat.head()
```

```python
bat = bat.drop(bat.index[0:101:2])
bat = bat.reset_index(drop = True)
bat.head()
```

```python
plt.plot(bat["Match_Date"], bat["Babar Azam"], label="Babar Azam", linestyle = "dashed", linewidth= 1, marker=".", color="blue")
plt.plot(bat["Match_Date"], bat["Mohammad Rizwan"], label="Rizwan", color="red")
plt.plot(bat["Match_Date"], bat["Shaheen Afridi"], label="Shaheen", color="yellow")
plt.legend()
plt.xticks(bat["Match_Date"][::2], rotation = 90)
plt.xlabel("dates")
plt.ylabel("scores")
plt.title("Babar vs Rizwan vs Shaheen")
plt.show()
```
