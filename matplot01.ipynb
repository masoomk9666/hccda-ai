```python
import seaborn as sns

tips = sns.load_dataset("tips")
tips.head()
```

```python
plt.scatter(tips["total_bill"], tips["tip"], c="red", s=20)
plt.show()
```

```python
plt.scatter(tips['total_bill'], tips['tip'], color ="blue", s=80)
plt.title("Total Bill vs Tip")
plt.xlabel("Total Bill")
plt.ylabel("Tip")
plt.show()
```

```python
plt.scatter(tips['total_bill'], tips['size'], color ="green", marker="^")
plt.title("Total Bill vs Table Size")
plt.xlabel("Total Bill")
plt.ylabel("Size")
plt.grid(True)
plt.show()
```

```python
plt.scatter(tips['tip'], tips['size'], s=tips["total_bill"])
plt.title("Tip vs Table Size")
plt.xlabel("Tip")
plt.ylabel("Size")
plt.grid(True)
plt.show()
```
