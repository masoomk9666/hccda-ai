```python
student={"ali":75,"sara":45, "Ahmed":60}
name=input("enter marks:")
for name in student:
    marks = student[name]
    print("Name:",name)
    print("Marks:", marks)

    if marks>=50:
        print("pass")
        break
    else:
        print("fail")

    print()
```
